# Scraping de Precios

Proyecto de web scraping listo para **Google Colab**.

**Pregunta:** ¿cómo extraer precios de una página web?

**Método:** BeautifulSoup + requests

**Stack:** requests, beautifulsoup4, pandas

Ejemplo con página de prueba.

## 0. Setup

```python
!pip install -q beautifulsoup4

import requests
from bs4 import BeautifulSoup
import pandas as pd
print("OK")
```

## 1. Hacer request

```python
# Ejemplo con una página simple (Books to Scrape)
url = "https://books.toscrape.com/"
headers = {"User-Agent": "Mozilla/5.0"}
response = requests.get(url, headers=headers)
print(response.status_code)
```

## 2. Parsear HTML

```python
soup = BeautifulSoup(response.content, "html.parser")
books = soup.find_all("article", class_="product_pod")
print(f"Libros encontrados: {len(books)}")
```

## 3. Extraer datos

```python
data = []
for book in books:
    title = book.h3.a["title"]
    price = book.find("p", class_="price_color").text
    data.append({"titulo": : title, "precio": : price})

df = pd.DataFrame(data)
print(df.head(10))
```

## 4. Guardar

```python
df.to_csv("precios.csv", index=False)
print("Guardado en precios.csv")
```

## 5. Conclusiones

- BeautifulSoup es ideal para sitios estáticos.
- Siempre respetá robots.txt y términos de servicio.
- Para sitios dinámicos usá Selenium o Playwright.

**Consejo:** agregá delays y rotación de User-Agents para scrapers más grandes.
